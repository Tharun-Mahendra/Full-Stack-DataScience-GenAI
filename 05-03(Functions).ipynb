### Global Keyword
-     The global keyword is used inside a function to indicate that it refers to global variable not local variable

```python
x=10
def update():
    x=x+6
    return x
```

```python
print(x)
update()
```

```python
x=10
def update():
    global x
    x=x+6
    return x
```

```python
print(x)
update()
```

### Globals()
-     globals() is a built-in function that returns a dictionary representing the current global symbol table.
-     It allows you to access and modify global variables programmatically.

```python
x=10
def update():
    globals()['x']=x+15 #acess & modify the global variable using dictionary
    return x
```

```python
print(x)
update()
```

### Passing List to a Function

```python
# Count of Even & Odd numbers in a list using Functions
def count(lst):
    even=0
    odd=0
    for i in lst:
        if i%2==0:
            even+= 1
        else:
            odd+= 1
    print('EvenCount:',even)
    print('OddCount:',odd)
```

```python
nums=[1,7,8,17,54,18,2,6,10]
count(nums)
```

```python
# Fibonnaci Sequence
def fib(n):
    a,b=0,1
    if n<=1:
        return a
    else:
        print(a)
        print(b)
        for i in range(0,n):
            c=a+b
            a=b
            b=c
            print(c)
```

```python
fib(8)
```

```python
# Factorial
def fact(n):
    res=1
    for i in range(1,n+1):
        res=res*i
    return res
```

```python
fact(5)
```

## Recursive Function
-     calling function itself

```python
def fun():
    print('hi')
fun()
```

```python
def fun():
    print('hi')
    fun()
```

```python
fun()
```

```python
import sys
sys.getrecursionlimit()
```

```python
sys.setrecursionlimit(45)
```

```python
i=0
def fun():
    global i
    i+=1
    print('hi',i)
    fun()
```

```python
fun()
```

```python
# factorial
def fact(n):
    if n==0:
        return 1
    else:
        return n*fact(n-1)
```

```python
fact(5)
```

## Anonymous or Lambda Function
-       a function without a name called as anonymous or lambda function

```python
# normal function
def square(n):
    return n*n
```

```python
square(5)
```

```python
# Lambda Function
s=lambda n:n*n #argument,operation
```

```python
s(5)
```

```python
add=lambda a,b:a+b
```

```python
add(10,55)
```
