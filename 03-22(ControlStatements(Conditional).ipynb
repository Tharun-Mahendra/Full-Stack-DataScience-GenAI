# If statement
- used to decide whether to execute a particular statement or not based on `condition`
- if condition is `True` it will Execute
### Syntax
-     if (condition):
- statement to execute
- condition is `true`

```python
if True:
    print('Condition executed')
    print('1st Statement')
print('output')
```

```python
if False:
    print('Condition executed')
print('no output')
```

```python
num=25
if(num>=15):
    print('num is greater than 15')
```

# If-else statement
- it wll execute the `if` block if condition is `true`
- `else` block if condition is `false`
### Syntax
-     if (condition):
- executes the block
- condition is `true`
-     else:
- executes the block
- condition is `false`

```python
if True:
    print('Condition Executed')
else:
    print('Not Executed')
print('executed based on condition true or false')
```

```python
if False:
    print('Condition Executed')
else:
    print('Condition Not Executed')
print('executed based on condition true or false')
```

```python
num=128456
if(num%2==0):
    print(num,"is even")
else:
    print(num,'is Odd')
```

```python
num=12485
if(num%2==0):
    print(num,'is Even')
else:
    print(num,'is Odd')
```

# Nested-if statement
- a statement inside the statement
### Syntax
-     if (condition1):
- executes when `condition1` is `true`
-     if (condition2):
- executes when `condition2` is `true`

```python
num=128456
if(num%2==0):
    print('Even')
    if (num>1000):
        print('Greater-Than 1000')
    else: 
        print('Less-Than 1000')
else:
    print('Odd')
```

# Elif statement
- multiple conditions in different statements
- as soon as one condition is `true` it comes out of block
### Syntax
-     if (condition):
- `statement`
-     elif(condition):
- `statement`
- `.`
- `.`
- `.`
-     else:
- `statement`

```python
num=10
if(num>0):
    print(num,'is Positive Number')
elif(num<0):
    print(num,'is Negative Number')
else:
    print('num is Zero')
```

```python
num=-25
if(num>0):
    print(num,'is Positive Number')
elif(num<0):
    print(num,'is Negative Number')
else:
    print('num is Zero')
```

```python
num=0
if(num>0):
    print(num,'is Positive Number')
elif(num<0):
    print(num,'is Negative Number')
else:
    print('num is Zero')
```

# Ternary Statement
- shortcut for using `if` and `if-else` statement
### if Shorthand 
- `if condition:statement`

```python
num=10
if num<25:print(num,'is less than 25')
```

### if-else Shorthand
- Statement when true `if` condition `else` Statement when false 

```python
num=10
print(num,'is Positive Number') if (num>=0) else print(num,'is Negative Number')
```

```python
num=-58
print(num,'is Positive Number') if (num>=0) else print(num,'is Negative Number')
```

## What are the different types of control statements in Python?  
#### In Python, control statements are used to alter the flow of execution based on specific conditions or looping requirements. The main types of control statements are: 
- `Conditional statements`: if, else, elif
- `Looping statements`: for, while
- `Control flow statements`: break, continue, pass, return
