#### Octal
- 0o indicates octal

```python
oct(10)
```

## Swappping two numbers

### Using temp variable

```python
a=10
b=20
```

```python
print(a,b)
```

```python
temp=a
a=b
b=temp
```

```python
print(a,b)
```

### Using operators

```python
a=10
b=20
```

```python
print(a,b)
```

```python
a=a+b #a=10+20=30
b=a-b #b=30-20=10
a=a-b #a=30-10=20
```

```python
print(a,b)
```

### Shortcut

```python
a=10
b=20
```

```python
print(a,b)
```
