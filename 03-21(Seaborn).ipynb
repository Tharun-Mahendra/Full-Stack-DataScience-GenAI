### univariate analysis
- working on single variable.

```python
plt.plot(df.InternetUsers) #using matplotlib
plt.show()
```

```python
sns.distplot(df.InternetUsers) #using seaborn
plt.show()
```

```python
sns.distplot((df.InternetUsers),bins=20) #bins means no.of bars by which we can visuall deeper
plt.show()
```

```python
sns.displot((df.InternetUsers),bins=20)
plt.show()
```

```python
sns.histplot((df.InternetUsers),bins=20)
plt.show()
```

```python
sns.boxplot(df.IncomeGroup)
plt.show()
```

### Bivariate analysis
- working on two variables

```python
sns.histplot((df.BirthRate,df.InternetUsers),bins=20)
plt.show()
```

```python
sns.boxenplot(data=df ,x='IncomeGroup', y='BirthRate')
plt.show()
```

```python
sns.boxplot(data=df ,x='IncomeGroup', y='BirthRate',color='r')
plt.show()
```

```python
sns.lmplot(df,x='InternetUsers',y='BirthRate')
plt.show()
```

```python
sns.lmplot(df,x='InternetUsers',y='BirthRate',fit_reg=False)
plt.show()
```

```python
sns.lmplot(df,x='InternetUsers',y='BirthRate',hue='IncomeGroup',fit_reg=False)
plt.show()
```

```python
sns.lmplot(df,x='InternetUsers',y='BirthRate',hue='IncomeGroup',fit_reg=False,height=8)
plt.show()
```

```python
sns.barplot(df,x='IncomeGroup',y='BirthRate',color='c')
plt.show()
```

```python
sns.catplot(df,x='IncomeGroup',y='BirthRate',color='k')
plt.show()
```

```python
sns.violinplot(df,x='IncomeGroup',y='BirthRate',color='r')
plt.show()
```

```python
sns.pointplot(df,x='IncomeGroup',y='BirthRate',color='y')
plt.show()
```
