# Bitwise Operators
- there are 6 types
- Complement ~
- And &
- Or |
- X-Or ^
- LeftShift <<
- RightShift >>

#### Complement ~
##### 1's Complement
- it will reverse the binary format of a given number and adds negative sign
- ex:12 binaryformat 00001100
- Complement -11110011
##### 2's Complement
- add 1 to 1's complement
- 11110011 + 00000001
- i.e 11110100 which is -12 in system memory  

```python
bin(12)
```

```python
~1100 #complement of 12
```

```python
-0b1101 #i.e -13
```

```python
bin(-45)
```

#### And &
- as we know if both the i/p's is true i.e 1&1 then only o/p is 1 else 0
- ex: 12&13 1100 & 1101 =1100 i.e 12

```python
1100 & 1101
```

```python
0b1100
```

```python
1 & 1 #i.e 0001&0001 =0001
```

```python
13 & 7 #i.e 1101&0111 =0101
```

```python
35 & 40
```

#### Or |
- same as and but if any one i/p is true i.e is 1 it gives o/p as true 1
- ex: 12|13 1100|1101 =1101 i.e 13

```python
12 | 13
```

```python
1 | 1
```

```python
13 | 7 #i.e 1101|0111 =1111 
```

```python
35 | 40
```

#### XOr ^ 
- combination of true & false gives o/p as 1 else 0
- ex: 12^13 1100^1101 =0001 i.e 1

```python
12 ^ 13
```

```python
1 ^ 1
```

```python
35 ^ 40
```

```python
13 ^ 7
```

#### LeftShift <<
- shifts the positions to left side we defaultly consider the binary number with zeros after decimal point which does effect value
- i.e 10->1010 as 1010.00000

```python
0b1010 << 2 #i.e shift 2 positions to left which becomes 1010.00000 to 101000.0000 i.e 40
```

```python
25<<1
```

```python
2<<1 
```

```python
5<<0 #as we are not shifting any 0's
```

#### RightShift >>
- same as leftshift but to right side

```python
0b1010>>2 #i.e shift 2 positions to rigth which becomes 1010.00000 to 10.100000 i.e 2
```

```python
25>>1
```

```python
2>>1
```

```python
5>>0
```
