# Split()
-     In Python, the split() method is used to divide a string into a list of substrings based on a specified delimiter.

```python
text = "Python is fun"
result = text.split()
print(result)
```

#### Splitting using a specific character

```python
text = "apple-banana-cherry"
result = text.split("-")
print(result)
```

#### Using max strings to split

```python
text = "Python is great and powerful"
result = text.split(" ",3)
print(result)
```

## Splitting date & time

```python
import pandas as pd
data = {'datetime': ['2025-03-26 & 14:15:00', '2025-03-27 & 09:30:00']}
df = pd.DataFrame(data)
df
```

```python
# Splitting date and time
df[['date', 'time']] = df['datetime'].str.split('&', expand=True)
df
```

```python
data = {'datetime': ['2025-03-26 14:15:00', '2025-03-27 09:30:00']}
df = pd.DataFrame(data)

# Convert the 'datetime' column to pandas datetime format
df['datetime'] = pd.to_datetime(df['datetime'])
# Extract individual components
df['year'] = df['datetime'].dt.year
df['month'] = df['datetime'].dt.month
df['day'] = df['datetime'].dt.day
df['hour'] = df['datetime'].dt.hour
df['minute'] = df['datetime'].dt.minute
df['second'] = df['datetime'].dt.second

df
```

### Splitting adress

```python
data = {'address': ['123 Main St, New York, NY', '456 High St, Chicago, IL']}
df = pd.DataFrame(data)
df
```

```python
# Splitting address into components
df[['street', 'city', 'state']] = df['address'].str.split(',', expand=True)
df
```
