# Python Functions
-     A Python function is a block of code that can be executed multiple times from different parts of a program. Functions allow you to:
- `Reuse code:` Write a block of code once and use it multiple times.
- `Improve code organization:` Break down large programs into smaller, manageable pieces.
- `Make code more readable:` Give a clear name to a block of code, making it easier to understand.
-     Basic Function Syntax:
*     def function_name(parameters):
*     code to be executed
*     return result
- `def` is the keyword to define a function.
- `function_name` is the name given to the function.
- `parameters` is a list of variables that are passed to the function.
- `return` result is the value returned by the function.

```python
def greet():
    print('Hello')
    print('How are you')
```

```python
greet()
```

```python
greet()
greet()
```

```python
greet('Tharun')
```

```python
# we have to declare argument to function to pass it when calling function
def greet(name):
    print('Hello')
    print('How are you',name)
```

```python
greet('Tharun')
```

```python
# adding two numbers using function
def add(x,y):
    return x+y
```

```python
add(5,6)
```

```python
# subtracting
def sub(x,y):
    return x-y
```

```python
sub(9,5)
```

```python
# lets do pass both functions together
def add_sub(x,y,z):
    return(x+y+z),(x-y-z)
```

```python
add_sub(5,4,8) # but it gives result as tuple
```

```python
x=add_sub(5,9,7)
print(x)
type(x)
```

```python
# lets make result in integer format
res1,res2=add_sub(8,4,7)
```

```python
print(res1)
type(res1)
```

```python
print(res2)
type(res2)
```
