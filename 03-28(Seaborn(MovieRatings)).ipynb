### Change datatype

## Building Dashboards

```python
f,axes = plt.subplots(2,2, figsize =(10,6))
sns.kdeplot(data=df,x='BudgetMillions',y='AudienceRating',ax=axes[0,0])
sns.kdeplot(data=df,x='BudgetMillions',y='CriticRating',ax=axes[0,1])

sns.violinplot(data=df,x='Genre',y='CriticRating',ax=axes[1,0])
sns.kdeplot(data=df,x='CriticRating',y='AudienceRating',ax=axes[1,1])
```

```python
sns.set_style('dark',{'axes.facecolor':'black'})
f, axes = plt.subplots (2,2, figsize = (15,15))

#plot [0,0]
sns.kdeplot(data=df,x='BudgetMillions',y='AudienceRating',shade=True, shade_lowest=True,cmap = 'inferno',ax = axes[0,0])
sns.kdeplot(data=df,x='BudgetMillions',y='AudienceRating',cmap='cool',ax = axes[0,0])

#plot [0,1]
sns.kdeplot(data=df,x='BudgetMillions',y='CriticRating',shade=True,shade_lowest=True, cmap='inferno',ax = axes[0,1])
sns.kdeplot(data=df,x='BudgetMillions',y='CriticRating',cmap='cool',ax=axes[0,1])

#plot[1,0]
sns.violinplot(data=df[df.Genre=='Drama'],x='YearOfRelease', y = 'CriticRating', ax=axes[1,0])

#plot[1,1]
sns.kdeplot(data=df,x='CriticRating',y='AudienceRating',shade = True,shade_lowest=False,cmap='Blues_r',ax=axes[1,1])
sns.kdeplot(data=df,x='CriticRating',y='AudienceRating',cmap='gist_gray_r',ax = axes[1,1])
```
